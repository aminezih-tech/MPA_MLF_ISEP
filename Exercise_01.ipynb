## Exercise 1.1 - Print Pattern

### Subtask:
Write a Python code that prints the given 'X' pattern using as few loops as possible, avoiding trivial solutions.


**Reasoning**:
The user wants to implement the logic for printing an 'X' pattern. I will provide a Python code block that follows the instructions: defining a size, using nested loops, and applying conditional logic for printing 'X' or ' '.



```python
size = 7 # Choosing an odd number for a symmetrical 'X' pattern

for row in range(size):
    for col in range(size):
        if row == col or row == size - 1 - col:
            print('X', end='')
        else:
            print(' ', end='')
    print()
```

**Reasoning**:

## Exercise 1.3 - Integer to Binary

### Subtask:
Write a Python script to convert an arbitrary integer to its binary representation as a string, without using the bin() function or other built-in Python conversion functions.


**Reasoning**:
The user wants a Python script to convert an integer to its binary representation. I will implement the `integer_to_binary` function following the detailed instructions, handling positive, negative, and zero inputs, and then test it with example cases.



```python
def integer_to_binary(number: int) -> str:
    if number == 0:
        return '0'

    is_negative = False
    if number < 0:
        is_negative = True
        number = abs(number)

    binary_digits = []
    while number > 0:
        remainder = number % 2
        binary_digits.append(str(remainder))
        number //= 2

    binary_string = ''.join(binary_digits[::-1])

    if is_negative:
        return '-' + binary_string
    else:
        return binary_string

# Test cases
print(f"Binary of 10: {integer_to_binary(10)}")
print(f"Binary of 0: {integer_to_binary(0)}")
print(f"Binary of -10: {integer_to_binary(-10)}")
print(f"Binary of 1: {integer_to_binary(1)}")
print(f"Binary of 42: {integer_to_binary(42)}")
print(f"Binary of -123: {integer_to_binary(-123)}")
```

## Exercise 1.4 - Fibonacci Sequence (Iterative)

### Subtask:
Implement the `fibonacci` function (iterative approach) to return a list that contains all Fibonacci numbers with values lower than the input integer.


**Reasoning**:
I will implement the `fibonacci` function using an iterative approach as described in the instructions, handling base cases and using a while loop to generate the sequence.



```python
def fibonacci(upper_threshold: int) -> list:
    fib_sequence = []

    if upper_threshold <= 0:
        return fib_sequence
    elif upper_threshold == 1:
        fib_sequence.append(0)
        return fib_sequence
    else:
        fib_sequence.append(0)
        fib_sequence.append(1)

    while True:
        next_fib = fib_sequence[-1] + fib_sequence[-2]
        if next_fib < upper_threshold:
            fib_sequence.append(next_fib)
        else:
            break

    return fib_sequence

# Test cases from the original problem description
print(fibonacci(10))
print(fibonacci(0))
print(fibonacci(1))
print(fibonacci(2))
print(fibonacci(100))
```

**Reasoning**:
Now I will implement the `fibonacci` function using a recursive approach, following the detailed instructions to define a helper recursive function and handle base cases for generating Fibonacci numbers up to the `upper_threshold`.



```python
def fibonacci_recursive(upper_threshold: int) -> list:
    fib_sequence = []

    if upper_threshold <= 0:
        return fib_sequence
    elif upper_threshold == 1:
        fib_sequence.append(0)
        return fib_sequence

    def _fib_recursive_helper(a: int, b: int):
        if a < upper_threshold:
            fib_sequence.append(a)

        next_val = a + b
        if next_val < upper_threshold:
            _fib_recursive_helper(b, next_val)

    _fib_recursive_helper(0, 1)
    return fib_sequence

# Test cases
print(f"Recursive fibonacci(10): {fibonacci_recursive(10)}")
print(f"Recursive fibonacci(0): {fibonacci_recursive(0)}")
print(f"Recursive fibonacci(1): {fibonacci_recursive(1)}")
print(f"Recursive fibonacci(2): {fibonacci_recursive(2)}")
print(f"Recursive fibonacci(100): {fibonacci_recursive(100)}")
```

## Exercise 1.4 - Fibonacci Sequence (Memoization)

### Subtask:
Implement the `fibonacci` function using a memoization approach to return a list that contains all Fibonacci numbers with values lower than the input integer.

#### Instructions:
1. Define a function `fibonacci_memoization(upper_threshold: int) -> list`.
2. Initialize an empty list `fib_sequence` to store the Fibonacci numbers.
3. Handle base cases: if `upper_threshold <= 0`, return `[]`. If `upper_threshold == 1`, return `[0]`.
4. Initialize a dictionary or list, `memo = {}`, to store computed Fibonacci numbers.
5. Define an inner helper function, for example, `_fib_memo(n: int)`, that calculates the nth Fibonacci number.
    a. If `n` is in `memo`, return `memo[n]`.
    b. If `n` is 0, return 0.
    c. If `n` is 1, return 1.
    d. Otherwise, compute `memo[n] = _fib_memo(n - 1) + _fib_memo(n - 2)` and return `memo[n]`.
6. Use a loop (e.g., `while True`) to generate Fibonacci numbers:
    a. Calculate the `current_fib` number using the `_fib_memo` helper function for an increasing index `i`.
    b. If `current_fib` is less than `upper_threshold`, append it to `fib_sequence`.
    c. Otherwise, break the loop.
7. Return the `fib_sequence` list.

**Reasoning**:
Now I will implement the `fibonacci_memoization` function following the detailed instructions, which involves using a dictionary for memoization within a helper recursive function to efficiently generate Fibonacci numbers up to the `upper_threshold`.



```python
def fibonacci_memoization(upper_threshold: int) -> list:
    fib_sequence = []
    memo = {}

    if upper_threshold <= 0:
        return fib_sequence
    elif upper_threshold == 1:
        fib_sequence.append(0)
        return fib_sequence

    def _fib_memo(n: int) -> int:
        if n in memo:
            return memo[n]
        if n == 0:
            return 0
        if n == 1:
            return 1

        result = _fib_memo(n - 1) + _fib_memo(n - 2)
        memo[n] = result
        return result

    i = 0
    while True:
        current_fib = _fib_memo(i)
        if current_fib < upper_threshold:
            fib_sequence.append(current_fib)
            i += 1
        else:
            break

    return fib_sequence

# Test cases
print(f"Memoization fibonacci(10): {fibonacci_memoization(10)}")
print(f"Memoization fibonacci(0): {fibonacci_memoization(0)}")
print(f"Memoization fibonacci(1): {fibonacci_memoization(1)}")
print(f"Memoization fibonacci(2): {fibonacci_memoization(2)}")
print(f"Memoization fibonacci(100): {fibonacci_memoization(100)}")
```

## Exercise 1.5 - Rock, Paper, Scissors (Basic)

### Subtask:
Implement the basic `rock_paper_scissors` function where a user plays one round against a random computer, printing the result. This will involve using the `random` library and `input()` function.


**Reasoning**:
I will implement the `rock_paper_scissors` function following the detailed instructions to simulate a basic one-round game against a random computer, handle user input, and determine the winner.



```python
import random

def rock_paper_scissors() -> None:
    choices = ['rock', 'paper', 'scissors']
    computer_choice = random.choice(choices)

    while True:
        user_input = input("Enter your choice (rock, paper, or scissors): ").lower()
        if user_input in choices:
            user_choice = user_input
            break
        else:
            print("Invalid choice. Please choose rock, paper, or scissors.")

    print(f"Computer chose: {computer_choice}")
    print(f"You chose: {user_choice}")

    if user_choice == computer_choice:
        print("It is a tie!")
    elif (
        (user_choice == 'rock' and computer_choice == 'scissors') or
        (user_choice == 'paper' and computer_choice == 'rock') or
        (user_choice == 'scissors' and computer_choice == 'paper')
    ):
        print("You win!")
    else:
        print("You lose!")

rock_paper_scissors()
```
